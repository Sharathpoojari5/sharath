# Namma Karkala: Kannada and Tulu voices (open-source models)

Makes a real voice recording for every line in `voice/script.json`, in **Kannada** and **Tulu**, with free open-source models:

* **Indic Parler-TTS** (`ai4bharat/indic-parler-tts`) for Kannada, one distinct voice per character.
* **Meta MMS-TTS Tulu** (`facebook/mms-tts-tcy`) for Tulu when it exists; otherwise the Kannada voice reads the Tulu text.

**Steps:** Runtime → Change runtime type → **T4 GPU** → Run all (about 15–30 minutes) → the zip downloads at the end → unzip it into `namma-karkala/voice/` so you get `voice/kn/*.mp3`, `voice/tcy/*.mp3` and `voice/manifest.json`.

Before you publish: read each model card's licence (MMS-TTS is non-commercial) and have a Tulu speaker proof-read the `tcy` lines in `voice/script.json`.
